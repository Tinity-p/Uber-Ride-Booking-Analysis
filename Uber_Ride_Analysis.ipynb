```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Date'].dtype
```

```python
df.info()
```

```python
df['Month'] = df['Date'].dt.month_name()
```

```python
df['Day'] = df['Date'].dt.day_name()
```

```python
df['Year'] = df['Date'].dt.year
```

```python
df['Time'] = pd.to_datetime(
    df['Time'],
    format='%H:%M:%S'
)

df['Hour'] = df['Time'].dt.hour
```

```python
len(df)
```

```python
df['Booking ID'].nunique()
```

```python
df['Booking Status'].value_counts()
```

```python
completed_rides = df[df['Booking Status'] == 'Completed']

len(completed_rides)
```

```python
cancelled = df[df['Booking Status'].str.contains(
    'Cancel',
    case=False,
    na=False
)]

cancellation_rate = len(cancelled) / len(df) * 100

cancellation_rate
```

```python
hourly_bookings = df.groupby('Hour').size()

print(hourly_bookings)
```

```python
hourly_bookings.plot(kind='line')

plt.xlabel("Hour")
plt.ylabel("Number of Bookings")
plt.title("Bookings by Hour")
plt.show()
```

```python
hourly_bookings.idxmax()
```

```python
daily_bookings = df['Day'].value_counts()

print(daily_bookings)
```

```python
daily_bookings.plot(kind='line')

plt.xlabel("Day")
plt.ylabel("Bookings")
plt.title("Bookings by Day")
plt.show()
```

```python
top_pickups = df['Pickup Location'].value_counts().head(10)

print(top_pickups)
```

```python
top_pickups.plot(kind='line')

plt.xlabel("Pickup Location")
plt.ylabel("Bookings")
plt.title("Top 10 Pickup Locations")
plt.show()
```

```python
vehicle_revenue = completed_rides.groupby('Vehicle Type')['Booking Value'].sum().sort_values(ascending=False)

vehicle_revenue
```
